```python
img_4 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\highlights_and_shadows.jpg')

# Convert BGR to RGB
rgb_img_4 = cv.cvtColor(img_4, cv.COLOR_BGR2RGB)

# Get image dimensions
h, w = rgb_img_4.shape[:2]

image_lab = cv.cvtColor(img_4, cv.COLOR_BGR2LAB)

# Split into L, a, b channels
L, a, b = cv.split(image_lab)

# Normalize L to [0, 1] range
L_normalized = L.astype(np.float32) / 255.0

# Apply gamma correction
gamma = 0.6
L_corrected = np.power(L_normalized, gamma) * 255.0
L_corrected = L_corrected.astype(np.uint8)

# Merge back and convert to BGR
corrected_lab = cv.merge([L_corrected, a, b])
corrected_image = cv.cvtColor(corrected_lab, cv.COLOR_LAB2BGR)

# Display the original and corrected images
fig, ax = plt.subplots(1, 2, figsize=(2 * w / 100, h / 100), dpi=100)

ax[0].imshow(rgb_img_4)
ax[0].set_title('Original')
ax[0].axis('off')

ax[1].imshow(cv.cvtColor(corrected_image, cv.COLOR_BGR2RGB))
ax[1].set_title(f'Gamma Corrected (γ={gamma})')
ax[1].axis('off')

plt.show()
```

```python
def histogram_equalization(image):
    hist, bins = np.histogram(image.flatten(), 256, [0, 256])

    # Calculate CDF
    cdf = hist.cumsum()
    cdf_normalized = cdf * hist.max() / cdf.max()  # For plotting

    # Normalize CDF to get the mapping function
    cdf_mapped = (cdf - cdf.min()) * 255 / (cdf.max() - cdf.min())
    cdf_mapped = cdf_mapped.astype('uint8')

    # Apply the mapping to the image
    equalized_image = cdf_mapped[image]

    return equalized_image, hist, cdf_normalized

image = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\shells.tif', cv.IMREAD_GRAYSCALE)

# Perform histogram equalization
equalized_img, original_hist, original_cdf = histogram_equalization(image)

# Compute histogram of the equalized image
equalized_hist, _ = np.histogram(equalized_img.flatten(), 256, [0, 256])

plt.figure(figsize=(12, 6))

# Original Histogram
plt.subplot(2, 2, 1)
plt.title("Original Image Histogram")
plt.hist(image.flatten(), 256, [0, 256], color='b')
plt.xlim([0, 256])

# Original CDF
plt.subplot(2, 2, 2)
plt.title("Original CDF")
plt.plot(original_cdf, color='r')
plt.xlim([0, 256])

# Equalized Histogram
plt.subplot(2, 2, 3)
plt.title("Equalized Image Histogram")
plt.hist(equalized_img.flatten(), 256, [0, 256], color='g')
plt.xlim([0, 256])

# Equalized CDF
plt.subplot(2, 2, 4)
plt.title("Equalized CDF")
plt.plot(np.cumsum(equalized_hist), color='m')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```

# Question 7

(a)  Sobel Filtering Using cv2.filter2D

```python
img_8 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\einstein.png', cv.IMREAD_GRAYSCALE)

# Define Sobel kernels
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])

sobel_y = np.array([[-1, -2, -1],
                    [ 0,  0,  0],
                    [ 1,  2,  1]])

# Apply Sobel using filter2D
grad_x = cv.filter2D(img_8, -1, sobel_x)
grad_y = cv.filter2D(img_8, -1, sobel_y)

# Compute gradient magnitude
gradient = np.sqrt(grad_x**2 + grad_y**2).astype(np.uint8)

# Display results
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(img_8, cmap='gray'), plt.title('Original')
plt.subplot(1, 3, 2), plt.imshow(grad_x, cmap='gray'), plt.title('Sobel X')
plt.subplot(1, 3, 3), plt.imshow(grad_y, cmap='gray'), plt.title('Sobel Y')
plt.show()
```

(b) Sobel Filtering Using manual convolution

```python
def manual_convolution(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    pad = kh // 2
    padded = np.pad(image, pad, mode='constant')
    output = np.zeros_like(image)

    for i in range(h):
        for j in range(w):
            patch = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(patch * kernel)

    return output

# Apply manual Sobel
grad_x_manual = manual_convolution(img_8, sobel_x)
grad_y_manual = manual_convolution(img_8, sobel_y)
gradient_manual = np.sqrt(grad_x_manual**2 + grad_y_manual**2).astype(np.uint8)

# Display manual results
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(grad_x_manual, cmap='gray'), plt.title('Manual Sobel X')
plt.subplot(1, 3, 2), plt.imshow(grad_y_manual, cmap='gray'), plt.title('Manual Sobel Y')
plt.subplot(1, 3, 3), plt.imshow(gradient_manual, cmap='gray'), plt.title('Gradient Magnitude')
plt.show()
```

(c) Sobel Filtering Using Separability

```python
# Define 1D kernels
kernel_v = np.array([[1], [2], [1]])  # Vertical smoothing
kernel_h = np.array([[1, 0, -1]])     # Horizontal derivative

# Apply separable convolution
grad_x_sep = cv.filter2D(cv.filter2D(img_8, -1, kernel_v), -1, kernel_h)
grad_y_sep = cv.filter2D(cv.filter2D(img_8, -1, kernel_h.T), -1, kernel_v.T)
gradient_sep = np.sqrt(grad_x_sep**2 + grad_y_sep**2).astype(np.uint8)

# Display separable results
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1), plt.imshow(grad_x_sep, cmap='gray'), plt.title('Separable Sobel X')
plt.subplot(1, 3, 2), plt.imshow(grad_y_sep, cmap='gray'), plt.title('Separable Sobel Y')
plt.subplot(1, 3, 3), plt.imshow(gradient_sep, cmap='gray'), plt.title('Gradient Magnitude')
plt.show()
```

```python
def zoom_nearest(img, s):
    h, w = img.shape[:2]
    new_h, new_w = int(h * s), int(w * s)
    zoomed = np.zeros((new_h, new_w, 3) if len(img.shape) == 3 else (new_h, new_w), dtype=img.dtype)

    for i in range(new_h):
        for j in range(new_w):
            orig_i = int(i / s)
            orig_j = int(j / s)
            zoomed[i, j] = img[orig_i, orig_j]

    return zoomed
```

```python
def zoom_bilinear(img, s):
    h, w = img.shape[:2]
    new_h, new_w = int(h * s), int(w * s)
    zoomed = np.zeros((new_h, new_w, 3) if len(img.shape) == 3 else (new_h, new_w), dtype=img.dtype)

    for i in range(new_h):
        for j in range(new_w):
            x = j / s
            y = i / s
            x1, y1 = int(x), int(y)
            x2, y2 = min(x1 + 1, w - 1), min(y1 + 1, h - 1)

            # Compute weights
            a = x - x1
            b = y - y1

            # Interpolate
            if len(img.shape) == 3:  # Color image
                for c in range(3):
                    top = (1 - a) * img[y1, x1, c] + a * img[y1, x2, c]
                    bottom = (1 - a) * img[y2, x1, c] + a * img[y2, x2, c]
                    zoomed[i, j, c] = (1 - b) * top + b * bottom
            else:  # Grayscale
                top = (1 - a) * img[y1, x1] + a * img[y1, x2]
                bottom = (1 - a) * img[y2, x1] + a * img[y2, x2]
                zoomed[i, j] = (1 - b) * top + b * bottom

    return zoomed
```

Compute Normalized SSD

```python
def normalized_ssd(original, zoomed):
    # Resize zoomed to match original dimensions
    zoomed_resized = cv.resize(zoomed, (original.shape[1], original.shape[0]))
    diff = original.astype(np.float32) - zoomed_resized.astype(np.float32)
    ssd = np.sum(diff ** 2)
    normalized_ssd = ssd / (original.shape[0] * original.shape[1])
    return normalized_ssd
```

# Question 9

(a) Using GrabCut to segment the image

```python
# Read the image
img_10 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\daisy.jpg')
img_10_rgb = cv.cvtColor(img_10, cv.COLOR_BGR2RGB)

# Create mask
mask = np.zeros(img_10.shape[:2], np.uint8)

# Allocate memory for models (required by grabCut)
bgdModel = np.zeros((1, 65), np.float64)
fgdModel = np.zeros((1, 65), np.float64)

# Rectangle around the flower
rect = (50, 100, img_10.shape[1]-50, img_10.shape[0]-300)

# Apply grabCut
cv.grabCut(img_10_rgb, mask, rect, bgdModel, fgdModel, 5, cv.GC_INIT_WITH_RECT)

# Create final mask where sure or probable foreground pixels are 1
mask2 = np.where((mask == 2) | (mask == 0), 0, 1).astype('uint8')

# Foreground and background
foreground = img_10_rgb * mask2[:, :, np.newaxis]
background = img_10_rgb * (1 - mask2[:, :, np.newaxis])

# Display results
plt.figure(figsize=(10, 8))
plt.subplot(1, 3, 1); plt.imshow(mask2, cmap='gray'); plt.title('Segmentation Mask')
plt.subplot(1, 3, 2); plt.imshow(foreground); plt.title('Foreground')
plt.subplot(1, 3, 3); plt.imshow(background); plt.title('Background')
plt.show()
```
