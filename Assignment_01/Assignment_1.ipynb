```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Question 1

```python
img_1 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\emma.jpg', cv.IMREAD_GRAYSCALE)

# Define control points
c = np.array([(50,50),(51,100),(149,255),(150,150),(255,255)])

# Create Transformation curve
t1 = np.linspace(0, c[0,1], c[0,0]+1-0)    #start, stop, num_of_points  <-- output
t2 = np.linspace(c[0,1]+1, c[1,1], c[1,0]-c[0,0])
t3 = np.linspace(c[1,1]+1, c[2,1], c[2,0]-c[1,0])
t4 = np.linspace(c[2,1], c[3,1], c[3,0]-c[2,0])
t5 = np.linspace(c[3,1]+1, c[4,1], c[4,0]-c[3,0])


# Combine Segments
t = np.concatenate((t1,t2,t3,t4,t5), axis=0).astype('uint8')

# Plot the transformation curve
fig ,ax = plt.subplots(2,1,figsize=(10,10))
ax[0].plot(t, 'b-', linewidth=2, label='Transformation Curve')
ax[0].scatter(c[:,0], c[:,1], color='red', s=50, label='Control Points')  # Mark control points
ax[0].set_xlabel('Input Pixel Intensity')
ax[0].set_ylabel('Output Pixel Intensity')
ax[0].set_title('Intensity Transformation Curve')
ax[0].grid(True)
ax[0].legend()
# plt.show()


# Apply Transformation
img_1_transformed = cv.LUT(img_1, t)

ax[1].imshow(img_1_transformed,cmap='gray')
plt.show()
```
