```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

# Question 1

```python
img_1 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\emma.jpg', cv.IMREAD_GRAYSCALE)

# Define control points
c = np.array([(50,50),(51,100),(149,255),(150,150),(255,255)])

# Create Transformation curve
t1 = np.linspace(0, c[0,1], c[0,0]+1-0)    #start, stop, num_of_points  <-- output
t2 = np.linspace(c[0,1]+1, c[1,1], c[1,0]-c[0,0])
t3 = np.linspace(c[1,1]+1, c[2,1], c[2,0]-c[1,0])
t4 = np.linspace(c[2,1], c[3,1], c[3,0]-c[2,0])
t5 = np.linspace(c[3,1]+1, c[4,1], c[4,0]-c[3,0])


# Combine Segments
t = np.concatenate((t1,t2,t3,t4,t5), axis=0).astype('uint8')

# Plot the transformation curve
fig ,ax = plt.subplots(2,1,figsize=(10,10))
ax[0].plot(t, 'b-', linewidth=2, label='Transformation Curve')
ax[0].scatter(c[:,0], c[:,1], color='red', s=50, label='Control Points')  # Mark control points
ax[0].set_xlabel('Input Pixel Intensity')
ax[0].set_ylabel('Output Pixel Intensity')
ax[0].set_title('Intensity Transformation Curve')
ax[0].grid(True)
ax[0].legend()
# plt.show()


# Apply Transformation
img_1_transformed = cv.LUT(img_1, t)

ax[1].imshow(img_1_transformed,cmap='gray')
plt.show()
```

```python
img_5 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\spider.png')
rgb_img_5 = cv.cvtColor(img_5, cv.COLOR_BGR2RGB)
hsv_img_5 = cv.cvtColor(img_5, cv.COLOR_BGR2HSV)

# Get image dimensions
h, w = map(float, rgb_img_5.shape[:2])


fig,ax = plt.subplots(1,2,figsize=(w/100,h/100), dpi=100)

# Show RGB (original colors)
ax[0].imshow(rgb_img_5, interpolation='nearest')
ax[0].set_title('RGB')
ax[0].axis('off')

# Show HSV properly — convert back to RGB for display
ax[1].imshow(cv.cvtColor(hsv_img_5, cv.COLOR_HSV2RGB), interpolation='nearest')
ax[1].set_title('HSV ')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
# Split into H, S, V
h, s, v = cv.split(hsv_img_5)

h_px, w_px = img_5.shape[:2]
dpi = 100
ncols = 3


# Parameters
sigma = 70
a = 0.75

# Create lookup table for all 256 possible pixel values
x_vals = np.arange(256)
transformed_vals = x_vals + a * 128 * np.exp(-((x_vals - 128)**2) / (2 * sigma**2))
transformed_vals = np.clip(transformed_vals, 0, 255).astype(np.uint8)

# Apply to saturation channel
s_transformed = cv.LUT(s, transformed_vals)

hsv_transformed = cv.merge([h, s_transformed, v])
bgr_transformed = cv.cvtColor(hsv_transformed, cv.COLOR_HSV2BGR)

fig, ax = plt.subplots(1, ncols, figsize=(ncols * w_px / dpi, h_px / dpi), dpi=dpi)



# Original image
ax[0].imshow(cv.cvtColor(img_5, cv.COLOR_BGR2RGB), interpolation='nearest')
ax[0].set_title('Original')
ax[0].axis('off')

# Enhanced image
ax[1].imshow(cv.cvtColor(bgr_transformed, cv.COLOR_BGR2RGB))
ax[1].set_title(f'Vibrance Enhanced (a={a})')
ax[1].axis('off')

# Transformation curve
ax[2].plot(x_vals, transformed_vals, color='red')
ax[2].set_title('Intensity Transformation')
ax[2].set_xlabel('Input Saturation')
ax[2].set_ylabel('Output Saturation')
ax[2].grid(True)

plt.tight_layout(pad=0.4)
plt.show()
```

```python
img_6 = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\shells.tif')

rgb_img_6 = cv.cvtColor(img_6, cv.COLOR_BGR2RGB)

h,w = img_6.shape[:2]
fig,ax = plt.subplots(1,1,figsize=(w/100,h/100), dpi=100)
ax.imshow(rgb_img_6)
ax.axis('off')
plt.show()
```

```python
def histogram_equalization(image):
    hist, bins = np.histogram(image.flatten(), 256, [0, 256])

    # Calculate CDF
    cdf = hist.cumsum()
    cdf_normalized = cdf * hist.max() / cdf.max()  # For plotting

    # Normalize CDF to get the mapping function
    cdf_mapped = (cdf - cdf.min()) * 255 / (cdf.max() - cdf.min())
    cdf_mapped = cdf_mapped.astype('uint8')

    # Apply the mapping to the image
    equalized_image = cdf_mapped[image]

    return equalized_image, hist, cdf_normalized

image = cv.imread(r'F:\SEM 5\Computer Vision\Project\Assignment_01\a1images\shells.tif', cv.IMREAD_GRAYSCALE)

# Perform histogram equalization
equalized_img, original_hist, original_cdf = histogram_equalization(image)

# Compute histogram of the equalized image
equalized_hist, _ = np.histogram(equalized_img.flatten(), 256, [0, 256])

plt.figure(figsize=(12, 6))

# Original Histogram
plt.subplot(2, 2, 1)
plt.title("Original Image Histogram")
plt.hist(image.flatten(), 256, [0, 256], color='b')
plt.xlim([0, 256])

# Original CDF
plt.subplot(2, 2, 2)
plt.title("Original CDF")
plt.plot(original_cdf, color='r')
plt.xlim([0, 256])

# Equalized Histogram
plt.subplot(2, 2, 3)
plt.title("Equalized Image Histogram")
plt.hist(equalized_img.flatten(), 256, [0, 256], color='g')
plt.xlim([0, 256])

# Equalized CDF
plt.subplot(2, 2, 4)
plt.title("Equalized CDF")
plt.plot(np.cumsum(equalized_hist), color='m')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```
